## 3. Recuento de viviendas y valores nulos

Calculamos el total de casas y buscamos qué columnas tienen valores nulos (`NaN`) y cuántos hay en cada una.

```python
num_viviendas = len(df)
print(f"Número total de viviendas en el dataset: {num_viviendas}")

# Sacamos las columnas con nulos
missing_series = df.isnull().sum()
missing_vars = missing_series[missing_series > 0].sort_values(ascending=False)

print("\nVariables con valores nulos (ordenadas de mayor a menor):")
print(missing_vars)
print(f"\nTotal de variables con datos nulos: {len(missing_vars)}")
```

### Comentarios sobre los nulos:
El dataset tiene 1460 casas y 19 columnas con valores nulos.

Mirando el archivo `data-description.txt`, se ve que en la mayoría de columnas el valor `NA` no es un fallo al recoger los datos, sino que indica que la casa no tiene ese elemento:
- `PoolQC` (1453 nulos), `MiscFeature` (1406), `Alley` (1369) y `Fence` (1179) tienen casi todo vacío porque la mayoría de las casas no tienen piscina, callejón o valla.
- En todas las columnas del garaje (`GarageType`, `GarageYrBlt`, `GarageFinish`, `GarageQual`, `GarageCond`) salen 81 nulos exactos, que son las 81 viviendas sin garaje.
- Con el sótano (`BsmtQual`, `BsmtCond`, etc.) pasa igual: esos 37-38 nulos son casas que no tienen sótano.
- En cambio, los nulos de `LotFrontage` (259) o `Electrical` (1) sí son datos que faltan de verdad y habrá que ver cómo rellenarlos luego.

```python
# Variables del punto 4
variables_estudio = [
    'Id', 'MSSubClass', 'MSZoning', 'LotFrontage', 'LotArea', 'Street', 'Alley',
    'LotShape', 'LandContour', 'Utilities', 'LandSlope', 'Neighborhood',
    'YearBuilt', 'YearRemodAdd',
    'GarageCars', 'GarageArea', 'GarageQual', 'GarageCond', 'GarageType', 'GarageYrBlt', 'GarageFinish',
    'OverallQual', 'OverallCond', 'SalePrice'
]

# Sacamos una tabla con los datos de cada una para verlas todas juntas
resumen_datos = []
for col in variables_estudio:
    n_nulos = df[col].isnull().sum()
    n_unicos = df[col].nunique()
    es_numerica = pd.api.types.is_numeric_dtype(df[col])
    
    val_min = df[col].min() if es_numerica else "N/A"
    val_max = df[col].max() if es_numerica else "N/A"
    
    resumen_datos.append({
        'Variable': col,
        'Tipo_Dato': str(df[col].dtype),
        'Nulos': n_nulos,
        'Valores_Únicos': n_unicos,
        'Mínimo': val_min,
        'Máximo': val_max
    })

df_resumen = pd.DataFrame(resumen_datos)
df_resumen
```

```python
# Quitamos las columnas que no nos aportan nada
df = df.drop(columns=['Id', 'Utilities'])

# Rellenamos con 'None' donde el nulo significa que la casa no tiene eso
df['Alley'] = df['Alley'].fillna('None')

garage_cat_cols = ['GarageType', 'GarageFinish', 'GarageQual', 'GarageCond']
for col in garage_cat_cols:
    df[col] = df[col].fillna('None')

# Si no tiene garaje le ponemos el año de construcción de la casa
df['GarageYrBlt'] = df['GarageYrBlt'].fillna(df['YearBuilt'])

# Rellenamos LotFrontage con la mediana de las casas de su mismo barrio
df['LotFrontage'] = df.groupby('Neighborhood')['LotFrontage'].transform(lambda x: x.fillna(x.median()))
```

## 4. Estudio individual de variables y preprocesamiento

Analizamos cada una de las 24 variables que pide el guión viendo su tipo, valores nulos, posibles outliers y qué hacemos con ellas para el modelado.

---

### Identificador y Parcela (`Id`, `LotFrontage`, `LotArea`)
- **`Id`:**
  - **Tipo:** Numérica discreta (es simplemente el identificador de la fila).
  - **Valores y nulos:** Va de 1 a 1460 y no tiene nulos.
  - **Decisión:** La eliminamos. No da ninguna información real sobre la casa y podría meter ruido en el modelo.

- **`LotFrontage` (Pies lineales de fachada a la calle):**

  - **Valores, nulos y outliers:** Va de 21.0 a 313.0 pies (mediana 69.0) y tiene 259 nulos. Hay algunos valores bastante altos por encima de 200 pies (outliers), llegando hasta 313.
  - **Tratamiento:** La mantenemos porque influye en el precio. Los nulos los rellenamos con la mediana de las casas de su mismo barrio (`Neighborhood`), que tiene más sentido que meter la media general de toda la ciudad.

- **`LotArea` (Superficie de la parcela en pies cuadrados):**

- **`MSZoning` (Clasificación de la zona):**
  - **Tipo:** Categórica nominal (`RL`, `RM`, `C (all)`, `FV`, `RH`).

- **`Street` (Tipo de acceso por carretera):**

  - **Valores y nulos:** 0 nulos. Casi todas son asfaltadas (`Pave`: 1454) y solo 6 son de grava (`Grvl`).

- **`Alley` (Tipo de acceso por callejón):**
  - **Tipo:** Categórica nominal (`Grvl`, `Pave`).
  - **Valores y nulos:** Tiene 1369 nulos.

## 7. Estudio de correlaciones (Clustermap)

Viendo la matriz de correlación y cómo agrupa el `clustermap` las variables numéricas, sacamos varios puntos importantes de cara a entrenar modelos:

---

### 1. Variables que más influyen en el precio (`SalePrice`):
Las que tienen una correlación positiva más fuerte con el precio de la vivienda son:
1. **`OverallQual` ($r = 0.79$):** Es la que más influye con diferencia. Cuanto mejor es la calidad de los materiales y acabados, más cara es la casa.
2. **`GrLivArea` ($r = 0.71$):** Los pies cuadrados habitables por encima del suelo. Lógicamente, a más superficie útil, mayor precio.
3. **Tamaño del garaje (`GarageCars` $r = 0.64$ y `GarageArea` $r = 0.62$):** La capacidad del garaje (tanto en número de coches como en superficie) tira bastante hacia arriba del precio.
4. **Sótano y primera planta (`TotalBsmtSF` $r = 0.61$ y `1stFlrSF` $r = 0.61$):** El tamaño de la base de la casa también tiene mucha relación con lo que cuesta.
5. **Baños y habitaciones (`FullBath` $r = 0.56$ y `TotRmsAbvGrd` $r = 0.53$):** El número de baños completos y el total de habitaciones.
6. **Año de construcción (`YearBuilt` $r = 0.52$ y `GarageYrBlt` $r = 0.51$):** Tienen correlación positiva porque las casas más nuevas se venden más caras que las antiguas.

---

### 2. Variables redundantes (colinealidad):
En los árboles del `clustermap` se ve muy claro que hay parejas de variables pegadas con colores muy rojos porque miden prácticamente lo mismo:
- **`GarageCars` y `GarageArea` ($r = 0.88$):** El número de coches que caben depende directamente de los metros del garaje. Para un modelo de regresión nos quedaríamos solo con `GarageCars` (que correlaciona un poco más con `SalePrice`) y quitaríamos `GarageArea` para no meter información repetida.
- **`YearBuilt` y `GarageYrBlt` ($r = 0.85$):** Casi siempre se construye el garaje a la vez que la casa (y más aún tras haber rellenado los nulos con `YearBuilt`).
- **`GrLivArea` y `TotRmsAbvGrd` ($r = 0.83$):** Cuantos más metros tiene la casa, más habitaciones salen.
- **`TotalBsmtSF` y `1stFlrSF` ($r = 0.82$):** El sótano suele ocupar la misma planta que el primer piso.

---

### 3. Variables que apenas tienen relación lineal con el precio:
- Columnas como el mes o el año en que se vendió la casa (`MoSold` con $r = 0.05$ y `YrSold` con $r = -0.03$) o `MiscVal` ($r = -0.02$) tienen una correlación casi de cero con `SalePrice`, por lo que por sí solas no sirven para predecir el precio linealmente.
- Como vimos en teoría, hay que tener en cuenta que correlación no implica causalidad, y que aquí estamos midiendo relación lineal (Pearson), por lo que podría haber relaciones no lineales que no se ven en esta matriz.
